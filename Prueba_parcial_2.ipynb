```python
def parteA():
  print("-- Parte A: Promedio de tres numeros --")
  numeros_string = input("Ingrese tres valores: ")
  numeros_list = numeros_string.split(',')
  numeros_clean_list = [i.strip() for i in numeros_list]
  numeros_list = [int(i) for i in numeros_clean_list]
  promedio = sum(numeros_list) / 3
  print(f"{promedio:.2f}")
  print()
```

```python
def parteB():
  print("-- Parte B: Directorio de contactos --")
  directorio = {}
  clave = ""
  while clave != "fin":
    clave = input("Ingrese el nombre (o 'fin' para terminar): ")
    if clave != "fin":
      valor = int(input("Ingrese el telefono: "))
      directorio[clave] = valor
  print("\nContactos guardados:")
  for i in directorio:
    print (i, "-", directorio[i])
  print()
```

```python
def menu():
  print("===== MENU PRINCIPAL =====")
  print("A. Calcular promedio de tres numeros")
  print("B. Crear un directorio de contactos")
  print("C. Salir")
  print()
```

```python
opcion = ""
while opcion != "C":
  menu()
  opcion = input("Seleccione una opcion: ").upper()
  print()
  if opcion == "A":
    parteA()
  elif opcion == "B":
    parteB()
  elif opcion == "C":
    print("Fin del programa. Hasta pronto!")
  else:
    print("Opcion invalida. Intentalo otra vez.")
```
