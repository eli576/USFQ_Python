```python
#boolean data
password = input("Do you have the password? (yes/no)").lower()
id = input("Do you have your student ID? (yes/no)").lower()

if password == "yes" and id == "yes":
  print("Access granted:", True)
else:
  print("Access granted:", False)
```

```python
#try/ except
inp = (input("Enter Fahrenheit Temp: "))
try:
  temp = float(inp)
  cel = (temp - 32.0) * 5.0/9.0
  print(cel)
except:
  print("please make sure you have entered a number")
```

```python

#try/ except task #2
# todo lo que puede causar un error lo pones entre el try

try:
  num = float(input("ingresa un numerador: "))
  den = float(input("ingresa un denominador: "))
  res = num/den
  if res > 1:
    print("the result is greater than 1")
  elif res < 1:
    print("the result is smaller than 1")
  else:
    print("the result is exaclty 1")

except:
  if den == 0:
    print("no se puede dividir por cero")
  else:
    print ("please enter a number")
```
