```python
precios = {
    "manzana" : 0.5,
    "banana" : 0.3,
    "naranja" : 0.4
}
```

```python
def multiplicar_precios (precios, factor):
  for key in precios:
    precios[key] *= factor
```

```python
factor = float(input("Ingrese el factor multiplicativo: "))
multiplicar_precios (precios, factor)
print(precios)
```
