```python
#Objetivo:
#Utilizar un lazo para recorrer un rango de números y sumar únicamente aquellos que sean pares
```

```python
#Pide al usuario dos números que definan el rango (inicio y fin).
inicio = int(input("Ingresa un numero entero para definir el inicio del rango: "))
fin = int(input("Ingresa un numero entero para definir el fin del rango: "))

#Recorre el rango e identifica los números pares.
pares = []
for i in range (inicio, fin+1):
  if i%2==0:
    pares.append(i)
print("Los numeros pares entre tu rango son:", pares)

#Acumula la suma de estos números y muestra el resultado.
suma = sum(pares)
print("La suma de los numeros pares entre tu rango es:", suma)
```
