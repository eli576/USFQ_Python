```python
def agregar(notas):
  nota = int(input("Ingresa una nota: "))
  notas.append(nota)
  print()
```

```python
def imprimir(notas):
  if notas != []:
    print("Notas:", notas)
  else:
    print("No hay notas")
  print()
```

```python
def vaciar(notas):
  notas.clear()
  print()
  # notas = [] no funciona proque crea nueva lista con el nombre igual
  # deberias escribir en la funcion main: notas = vaciar(notas)
```

```python
def eliminar_ult(notas):
  notas = notas.pop()
  print()
```

```python
def nota_baja_alta(notas):
  nota_max = max(notas)
  nota_min = min(notas)
  print("La nota mas alta es:", nota_max)
  print("La nota mas baja es:", nota_min)
  print()
```

```python
def histograma(notas):
  for nota in notas:
    bar = "*"*nota
    print(f"{nota:2}:{bar}") #:2 imprima para que los asteriscos empiezen en la misma linea.
  print()
```

```python
def menu():
  print("Menu:")
  print("1: Agregar notas")
  print("2: Imprimir notas")
  print("3: Vaciar notas")
  print("4: Eliminar la ultima nota")
  print("5: Encontrar la nota mas alta y la mas baja")
  print("6: Histograma de notas")
  print()
```

```python
def main():
  notas = []
  opcion = 0
  while opcion != 7:
    menu()
    opcion = int(input("Elige una opcion: "))
    print()
    if opcion == 1:
      agregar(notas)
    elif opcion == 2:
      imprimir(notas)
    elif opcion == 3:
      vaciar(notas)
    elif opcion == 4:
      eliminar_ult(notas)
    elif opcion == 5:
      nota_baja_alta(notas)
    elif opcion == 6:
      histograma(notas)
    elif opcion == 7:
      histograma_vertical(notas)
    elif opcion == 7:
      print("Adios!")

```
