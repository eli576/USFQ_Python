```python
import random
```

```python
#Crea una lista llamada calificaciones con 8 números enteros entre 0 y 100, que representen las calificaciones de un grupo de estudiantes.
calificaciones = []
for num in range(8):
  calificaciones.append(random.randint(0,100))
#Imprimir cada calificación.
print("Calificaciones en la lista: ", calificaciones)
```

```python
#Usa un lazo for para recorrer la lista y:
#Contar cuántas calificaciones son aprobatorias (≥ 60) y cuántas son reprobatorias.
aprobatorias = 0
reprobatorias = 0
for num in calificaciones:
  if num >= 60:
    aprobatorias = aprobatorias + 1
  else:
    reprobatorias = reprobatorias + 1
print("Número de calificaciones aprobatorias: ", aprobatorias)
print("Número de calificaciones reprobatorias: ", reprobatorias)
```

```python
#Calcular el promedio de las calificaciones.
promedio = sum(calificaciones)/8
print ("Promedio de la clase: ", promedio)
```

```python
#Finalmente, pide al usuario una calificación y verifica si está en la lista. Si está, muestra un mensaje indicando que la calificación existe; si no, indica que no está en la lista
busca = int(input("Ingrese una calificación para buscar: "))
if busca in calificaciones:
  print(f"La calificación {busca} está en la lista")
else:
  print(f"La calificación {busca} no está en la lista")
```
