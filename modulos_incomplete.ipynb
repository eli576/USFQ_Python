```python
module_path = '/content/drive/MyDrive/My_Python_Modules/nasa_data.py'
with open(module_path, 'w') as f:
  f.write("""
from space_utils import planets, distances

def dictionary_planets(planets, distances):
  planet_distances = dict(zip(planets, distances))
  return planet_distances""")
```

```python
import space_utils
space_utils.mission_countdown(10)

import nasa_data

# Call dictionary_planets without arguments as it now imports them internally
planet_distances = nasa_data.dictionary_planets(planets, distances)

search = input("What planet do you want to look up?: ").lower()

found = False
for key in planet_distances:
  if search == key: # `planets` in space_utils are lowercase, so direct comparison works
    print (f"The distance of {key.capitalize()} from the sun is {planet_distances[key]} million km.")
    found = True
    break

if not found:
  print(f"Sorry, I could not find information for '{search}'.")
```
