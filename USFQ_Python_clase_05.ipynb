```python
n = 5
while n>0: #condicion logica
  print (n)
  n = n-1
print("Blastoff!")
```

```python
n = 5
while n>0:
  print (n, end = ' ') # end = '' para que todo se imprima al lado
  n = n-1
print("Blastoff!")
```

```python
while True: # algo hasta el infinito
  line = input('> ')
  if line == 'stop': # para detenerlo
    break
  print(line)
print('Loop stopped')
```

```python
while True:
  line = input('> ')
  if line[0] == '#':
    continue # no ejectues nada de que estoy haciendo abajo - termina la iteracion actual y se va a la proxima iteracion
  if line == 'done':
    break # termina el loop (bucle)
  print(line)
print('Done!')
```

```python
n = 0
while n<=100:
  n=n+1
  if n%7 != 0:
    continue
  print (n, end = ' ')
print('End')
```

```python
while True:
  ph = float(input("Enter a pH value: "))
  if ph < 0 or ph > 14:
    print("Warning: invalid pH value. Skipping...")
    continue
  if ph == 7:
    print("Neutral solution detected")
  elif ph < 3 or ph> 11:
    print("Highly reactive substance! Stopping measurement")
    break
  print("Measurement recorded: ", ph)
```
