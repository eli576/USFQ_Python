```python
import random
```

```python
#2) Generar 24 números enteros pseudoaleatorios
#3) Imprimir los 24 números
numeros = []
for num in range(24):
  numeros.append(random.randint(-1000,1000))
print(numeros)
```

```python
#4) Encontrar el número máximo de los mismos 24 números; usar el lazo “for” presentado en clases
max = numeros[0]
for num in numeros:
  if num > max:
      max = num
  else:
    pass
print (max)
```

```python
#5) Encontrar el número mínimo de los mismos 24 números; usar el lazo “for” presentado en clases
min = numeros[0]
for num in numeros:
  if num < min:
      min = num
  else:
    pass
print (min)
```
