```python
#Objetivo: Crear un programa que permita al usuario gestionar una lista de compras. Usen funciones en cada caso.
```

```python
#Agregar un artículo
def agregar(articulos):
  articulo = input("Qué quieres comprar?: ").lower()
  articulos.append(articulo)
  print(f"'{articulo}' ha sido añadido a la lista.")
  print()
```

```python
#Mostrar la lista
def mostrar(articulos):
  if articulos != []:
    print("Lista de compras:", articulos)
  else:
    print("La lista de compras está vacía")
  print()
```

```python
#Eliminar un artículo por nombre
def eliminar_nombre(articulos):
  eliminacion = input("Cual artículo quieres eliminar?: ").lower()
  if eliminacion in articulos:
    articulos.remove(eliminacion)
    print(f"'{eliminacion}' ha sido eliminado de la lista.")
  else:
    print(f"'{eliminacion}' no se encuentra en la lista.")
  print()
```

```python
#Eliminar el último artículo agregado
def eliminar_ult(articulos):
  articulos = articulos.pop()
  print(f"El útlimo artículo ha sido eliminado de la lista.")
  print()
```

```python
#Buscar si un artículo está en la lista
def buscar(articulos):
  busca = input("Qué artículo estás buscando?: ").lower()
  if busca in articulos:
    print("El artículo está en la lista")
  else:
    print("El artículo no está en la lista")
  print()
```

```python
#Vaciar la lista
def vaciar(articulos):
  articulos.clear()
  print("La lista está vacía.")
  print()
```

```python
def menu():
  print("Menu:")
  print("1: Agregar artículo")
  print("2: Mostrar lista de compras")
  print("3: Eliminar artículo")
  print("4: Eliminar el último artículo")
  print("5: Buscar artículo")
  print("6: Vaciar lista de compras")
  print()
```

```python
def main():
  articulos = []
  opcion = 0
  while opcion != 7:
    menu()
    opcion = int(input("Elige una opcion: "))
    print()
    if opcion == 1:
      agregar(articulos)
    elif opcion == 2:
      mostrar(articulos)
    elif opcion == 3:
      eliminar_nombre(articulos)
    elif opcion == 4:
      eliminar_ult(articulos)
    elif opcion == 5:
      buscar(articulos)
    elif opcion == 6:
      vaciar(articulos)
    elif opcion == 7:
      print("Adios!")
```

```python
main()
```
