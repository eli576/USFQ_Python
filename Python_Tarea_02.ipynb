```python
def verifica_tamano_mensaje(mensaje):
  if len(mensaje) < 5:
    return False
  else:
    return True
```

```python
def analizar_mensaje(mensaje, positivas, negativas):
  suma_positivas = 0
  suma_negativas = 0
  palabras_mensaje = mensaje.lower().split()
  for palabra in palabras_mensaje:
    if palabra in positivas:
      suma_positivas += 1
    elif palabra in negativas:
      suma_negativas += 1

  if suma_positivas > suma_negativas:
    sentimiento = "positivo"
  elif suma_positivas < suma_negativas:
    sentimiento = "negativo"
  else:
    sentimiento = "neutral"
  return sentimiento
```

```python
# Mensajes

m1 = "Hoy fue un día excelente"
m2 = "Estoy muy triste con los resultados"
m3 = "Hola que haces"
m4 = "Odio levantarme temprano"
m5 = "Gracias por el apoyo"
m6 = "ok"

mensajes = [m1, m2, m3, m4, m5, m6]

#Palabras clave
positivas = ["excelente", "feliz", "bueno", "gracias", "increíble", "bien"]
negativas = ["malo", "triste", "terrible", "odio", "aburrido", "horrible"]

```

```python
def main():
  positivos = 0
  negativos = 0
  neutrales = 0
  no_analizados = 0

  for i, mensaje in enumerate(mensajes, start=1):
    if verifica_tamano_mensaje(mensaje) == True:
      sentimiento = analizar_mensaje(mensaje, positivas, negativas)
      print(f"Mensaje {i}: {sentimiento}")
      if sentimiento == "positivo":
        positivos += 1
      elif sentimiento == "negativo":
        negativos += 1
      else:
        neutrales += 1
    else:
      no_analizados += 1
      print(f"Mensaje {i}: muy corto, no se analiza")

  print(f"\nResumen:\nPositivos: {positivos}\nNegativos: {negativos}\nNeutrales: {neutrales}\nMensajes no analizados: {no_analizados}")
```

```python
main()
```
