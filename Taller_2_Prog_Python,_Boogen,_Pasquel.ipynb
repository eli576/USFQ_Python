```python
#Ejercicio 4
word = str(input("Ingrese dos palabras: ")).lower()

#Salida 1
word_split = word.split()
segunda_palabra = word_split[1]
print(f"Salida 1: {segunda_palabra}")

#Salida 2
primera_palabra = word_split[0]
palabra_alternando = ""
for i in range(len(primera_palabra)):
 if i % 2 == 0:
  palabra_alternando += primera_palabra[i].upper()
 else:
  palabra_alternando += primera_palabra[i].lower()
print(f"Salida 2: {palabra_alternando}")

#Salida 3
print(f"Salida 3: {segunda_palabra[::-1]}")

#Salida 4
vocales = "aeiou"
unica = set()

for letra in word:
    if letra in vocales:
        unica.add(letra)

print(f"Salida 4: En las palabras aparecen {len(unica)} vocales diferentes.")

#Salida 5
counts = {}
for letra in word:
    if letra in counts:
        counts[letra] += 1
    else:
        counts[letra] = 1

if counts:
    max_count = 0
    for letra in counts:
        if counts[letra] > max_count:
            max_count = counts[letra]

    letra_frec = []
    for letra, count in counts.items():
        if count == max_count:
            letra_frec.append(letra)

    print(f"Salida 5:La/s letra/s más frecuente/s en la cadena combinada es/son: {', '.join(letra_frec)} con {max_count} apariciones.")
else:
    print("No hay letras para analizar en la cadena combinada.")

```

```python
#Ejercicio 5
puntosX = [1 ,2 ,3 ,4 ,5]
puntosY_real = [2, 2.5, 2.9, 1.7, 5.2]
m = 0.8
b = 1.2
puntosY_estimados = []

for x in puntosX:
  y_estimado = ((m*x) + b)
  puntosY_estimados.append(y_estimado)
print("Puntos Y estimados:", puntosY_estimados)

sumador = []
for i in range(len(puntosY_real)):
  resultado = abs(puntosY_real[i] - puntosY_estimados[i])
  sumador.append(resultado)
print("Diferencia entre puntos Y estimados y reales:", sumador)

suma = sum(sumador)

print("Suma:", suma)

ECM = (1/len(puntosY_real))*suma

print("ECM:", ECM)
```
