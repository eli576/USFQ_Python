```python
productos = int(input("Ingresa el numero de productos para inspeccionar: "))
inspections = int(input("Ingresa el numero de etapas de inspeccion: "))

for x in range(1, productos + 1): # +1 para asegurar que va al ultimo numero
  for i in range (inspections, 0, -1): #en reverso
    print(f"Inspecting Product {x} - Stage {i}")
```
