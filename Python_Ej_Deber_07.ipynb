```python
#Implementa dos funciones para identificar los números primos en un rango definido por el usuario.
```

```python
#Función es_primo(num):
#Debe recibir un número entero y retornar True si es primo o False si no lo es.

def es_primo(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True
```

```python
#Función primos_en_rango(inicio, fin):
#Recibe dos números enteros que definen un rango (inclusive).
#Utiliza es_primo para generar y retornar una lista con todos los números primos dentro de ese rango.

def primos_en_rango(inicio, fin):
  lista = []
  for i in range(inicio, fin+1):
    if es_primo(i):
      lista.append(i)
  print(f"Los numeros primos entre {inicio} y {fin} son: {lista}")
  return lista
```

```python
#Programa Principal:
#Solicita al usuario dos números que representen el inicio y el fin del rango.
#Valida que el valor de inicio sea menor o igual que el fin.
#Imprime la lista de números primos encontrados.

inicio = int(input("Ingresa el primer numero del rango: "))
fin = int(input("Ingresa el ultimo numero del rango: "))
if inicio > fin:
  print("El primer numero debe ser menor o igual que el ultimo")
else:
  primos_en_rango(inicio,fin)
```
