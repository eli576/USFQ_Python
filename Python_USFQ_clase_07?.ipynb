```python
lista = [75, 23, 8, 473, 4, 321, 62, 15]

min = lista[0]
max = lista[0]
for i in lista:
  if i < min:
    min = i
  if i > max:
    max = i

print(f"min = {min}, max = {max}")
```

```python
lista = []
while True:
  num = int(input("Ingresa un numero entero (>2): "))
  if num > 2:
    prime = True
    for i in range (0, num + 1):
      for j in range (2, i):
        if i%j == 0:
          prime = False
          break
        if prime = True:
          lista.append(i)

    print(f"Numeros primo hasta {num}: {lista}")
  else:
    print("No hay numeros primo debajo de 2")
  key = str(input("Would you like to try again? (y/n): "))
  if key == "n":
    break
```
