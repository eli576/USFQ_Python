```python
import random
import math
```

```python
distances = []
for i in range(1,1000000):
  dice_a = random.randint(1,6)
  dice_b = random.randint(1,6)
  distance = math.fabs(dice_a - dice_b)
  distances.append(distance)
  #print(f"Roll {i}: Dice A = {dice_a}, Dice B = {dice_b} -> Distance = {distance}")

Average = sum(distances) / len(distances)
print(f"Average distance: {Average}")
```

```python
def calculate_ticket_price (age, is_student):
  if age < 18:
    ticket_price = 8
  else:
    ticket_price = 10
  if is_student == "yes":
    ticket_price = ticket_price - ((ticket_price/100)*20)
  return ticket_price
```

```python
def print_receipt(name, age, is_student):
  ticket_price = calculate_ticket_price (age, is_student) # Capture the returned value
  print(f"  --- Cinema Receipt --- \nName: {name} \nAge: {age} \nStudent: {is_student} \nFinal ticket price: {ticket_price} USD")
  print()
```

```python
def es_negativo(numero):
  if numero <= 0:
    return True
  else:
    return False
```

```python
def logaritmo_serie(a, n_terminos):
  if es_negativo(a) == False:
    x = (a-1)/(a+1)
    accum = 0
    for n in range(0, n_terminos+1):
    # or instead do con salto 2 range(0, n_terminos, 2)
    # ln(a) = 2 * (x + x³/3 + x⁵/5 + x⁷/7 + ...)
    #term = x**n/n
      term = (x**(2*n + 1)) / (2*n + 1)
      accum += term
    ln = 2*accum
    return ln
  else:
    print("Error: el numero debe ser positivo")
    return None
```

```python
a = int(input("Ingresa un numero para calcular su logaritmo natural: "))
n_terminos = int(input("Ingresa el numero de terminos: "))
logaritmo_serie(a, n_terminos)
```
