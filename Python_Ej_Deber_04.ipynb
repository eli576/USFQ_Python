```python
#Escribe un programa en Python que permita al usuario ingresar una cantidad indefinida de números enteros (positivos y negativos)
#El programa debe terminar cuando el usuario ingrese el número 0 y luego mostrar un análisis de los datos ingresados.

#Requisitos del programa:
#Usar un bucle while para permitir que el usuario ingrese números hasta que escriba 0 para terminar.
#Al finalizar, el programa debe mostrar:
#La cantidad de números ingresados (sin contar el 0).
#La suma total de los números.
#El promedio de los números ingresados.
#El número mayor y el número menor ingresados.
#La cantidad de números positivos y negativos ingresados.
```

```python
numeros = []
numero = []
cantidad_positivos = 0
cantidad_negativos = 0

while numero != 0:
  numero = int(input("Ingrese un número (0 para terminar): "))
  if numero != 0:
    numeros.append(numero)
    if numero >  0:
      cantidad_positivos = cantidad_positivos + 1
    else:
      cantidad_negativos = cantidad_negativos + 1

cantidad = len(numeros)
suma = sum(numeros)
promedio = suma / cantidad
numero_mayor = max(numeros)
numero_menor = min(numeros)

print("\nCantidad de números ingresados:", cantidad)
print("Suma total:", suma)
print("Promedio de los números ingresados:", promedio)
print("Número mayor:", numero_mayor)
print("Número menor:", numero_menor)
print("Cantidad de positivos:", cantidad_positivos)
print("Cantidad de negativos:", cantidad_negativos)
```
