```python
#Ejercicio diccionarios - analisis de sentimientos
```

```python
tweet_1 = "Happy happy crowd at the concert tonight!"
tweet_2 = "What a happy night, so much joy and fun!"
tweet_3 = "Such a happy vibe, happy faces everywhere!"
```

```python
#1. limpiar tweets
def clean_string(tweet):
  clean_tweet = tweet.replace('!', '').replace(',', '').lower()
  return clean_tweet
```

```python
tweet_1 = clean_string(tweet_1)
tweet_2 = clean_string(tweet_2)
tweet_3 = clean_string(tweet_3)
```

```python
#2. combinar tweets en una sola lista
all_tweets = tweet_1 + ' ' + tweet_2 + ' ' + tweet_3
print(all_tweets)
```

```python
#3. transformar string en lista
lista = all_tweets.split()
print(lista)
```

```python
counts = {}

for word in lista:
  if word not in counts:
    counts[word] = 1
  else:
    counts[word] += 1
print(counts)
```

```python
# alternativa (lo mismo como arriba)
for word in counts:
  print(word, counts[word])
```
