## Selección Sort

El algoritmo de selección (selection sort) es un algoritmo de ordenamiento simple que funciona de la siguiente manera:

1. Encuentra el elemento mínimo de la lista no ordenada.
2. Intercambia este elemento mínimo con el primer elemento no ordenado.
3. Repite los pasos 1 y 2 para el resto de la lista no ordenada hasta que toda la lista esté ordenada.

```python
#3. Ordenar dicha lista utilizando el algoritmo de selección.
n = len(lista)
for i in range(n):
  min = i
  for j in range(i+1, n):
    if lista[j] < lista[min]:
      min = j
  lista[i], lista[min] = lista[min], lista[i]
```

```python
#4. Mostrar la lista ya ordenada.
print(lista)
```
