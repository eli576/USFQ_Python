```python
edad = int(input("Edad: "))
if edad < 12:
  precio = 3
elif edad >= 12 and edad <= 17:
  precio = 5
elif edad >= 18 and edad <= 59:
  precio = 8
else:
  precio = 4

dia = str(input("Día: ")).lower()
if dia == "miercoles" or dia == "miércoles":
  precio = precio - 1
else:
  pass

estudiante = str(input("¿Estudiante? (s/n): ")).lower()
if estudiante == "s":
  precio = precio - 1
else:
  pass

funcion_3d = str(input("¿Función 3D? (s/n): ")).lower()
if funcion_3d == "s":
  precio = precio + 2
else:
  pass

print(f"Precio final: ${precio}")
```
